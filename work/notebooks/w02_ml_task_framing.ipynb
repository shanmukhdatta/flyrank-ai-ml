# ML-03 — Frame Your Lane as an ML Task

## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

### My Lane as an ML Task

>I frame this project as a **supervised prediction (classification)** task. The goal is to use historical search, engagement, and content signals to predict whether a webpage is likely to **grow, decline, or recover** during a future time window. Although the project involves time-series feature engineering, the final objective is to classify future page performance so that content teams can prioritize pages for review.


## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

### Target or Proxy

>The target will be a **future observed outcome** rather than a current rule. I plan to define the target by comparing a future time window with a previous feature window. For example, I may use the previous 90 days of search and engagement signals to predict whether a webpage experiences growth, decline, or recovery during the following 30 days.

>This target is created from future observed performance rather than copied from an existing product score or rule. Using future outcomes helps reduce the risk of circular reasoning and better reflects the real decision problem.


## 3. Success metric

*One metric you can defend. What number means 'good'?*

### Success Metric

>The primary evaluation metric will be **Precision@K**, where **K** represents the number of pages that a content team can realistically review. Since the goal is to prioritize the most important pages rather than classify every page perfectly, Precision@K better reflects the real business decision. I will also report supporting metrics such as Precision, Recall, F1-score, and ROC-AUC to provide a more complete evaluation of model performance.


## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

### Unit of Analysis

>The unit of analysis in this project is **one webpage over a defined historical feature window**. Each training example represents a single content item with features engineered from its historical daily search and engagement data. The target represents the webpage's future performance during the following prediction window.

>This structure allows the model to learn temporal patterns while keeping the prediction process free from future information leakage.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

### Why Machine Learning Instead of a Fixed Rule?

>A fixed rule can only capture simple conditions such as "traffic decreased by 20%" or "CTR is below a threshold." However, webpage performance is influenced by many interacting factors, including search visibility, engagement, content freshness, ranking position, seasonality, and historical trends. These relationships are often nonlinear and vary across different pages.

>Machine learning can learn complex interactions between these signals and assign different importance to each feature based on historical evidence. Rather than replacing human judgment, the model provides a ranked estimate of future growth, decline, or recovery that helps prioritize content review more effectively than a single manually defined rule.
