## 2. The question: decision, action, cost of a wrong call


### Research Question

>**Research Question:** Can historical search and engagement signals be used to predict which webpages are likely to decline, recover, or gain momentum over a future time window?

>The decision my project aims to improve is **which webpages should be reviewed first** before their performance changes significantly. The primary users of this prediction are SEO analysts, content strategists, and marketing teams who have limited time and resources to review thousands of pages.

>A correct prediction allows the team to prioritize high-impact pages early, potentially preventing traffic loss or supporting content that is likely to grow. A wrong recommendation may cause valuable pages to be ignored or lead the team to spend time reviewing pages that were unlikely to change. Therefore, the model should support human decision-making rather than replace human judgment.

## 4. Careful words: what I can and can't claim

### What My Project Can and Cannot Claim

>This project is a **decision-support** system based on observed historical search and engagement signals. It can identify patterns associated with future growth, decline, or recovery and provide ranked recommendations that help prioritize content review. The results are observational and predictive rather than causal.

>This project **cannot** prove that updating a page will increase traffic, determine how Google's ranking algorithms work, or guarantee future performance. The model only estimates likely future outcomes using the information available before the prediction period, and all predictions should be interpreted as probabilistic guidance that supports human decision-making rather than replacing it.
